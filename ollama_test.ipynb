```python
import av
import cv2


def first_n_frames(path: str, n: int):
    frames = []
    with av.open(path) as container:
        stream = container.streams.video[0]
        for packet in container.demux(stream):
            for frame in packet.decode():
                img = frame.to_ndarray(format="bgr24")  # NumPy array
                ok, buf = cv2.imencode(".jpg", img, [cv2.IMWRITE_JPEG_QUALITY, 90])
                frames.append(buf.tobytes())
                if len(frames) == n:
                    return frames
    return frames
```

```python
video_bytes = first_n_frames("filestore/f88f6ee8-e072-45dd-8b10-7ee45f3c7083.mp4", 200)
```

```python
with open("filestore/f88f6ee8-e072-45dd-8b10-7ee45f3c7083.en.vtt", "r") as f:
    captions = f.read()

response = client.generate(
    model="gemma3:27b",
    prompt=f"Explain what is happening in this video. I am providing the captions, but focus on what is happening in the video. Do not ask any follow up questions. This is the start of the video, but the captions for the entire video are: {captions}",
    images=video_bytes,
)
# print(response.response)
from IPython.display import display, Markdown

display(Markdown(response.response))
```
